```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
numbers = np.arange(100)

squares = numbers * numbers
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(numbers, squares, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
```

```python
plt.scatter(x_train, y_train)
```

```python
# y = b0 + b1X  (linear relation - monomial)
# b0 = bias
# b1 = weight

Degree greater than 1
y = b0 + b1X + b2X2  (polynomial relation)
y = b0 +  b1X + b2X2 + b3X3

y = 2X (weight) + 0
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_reg = PolynomialFeatures(degree = 2)

x_poly = poly_reg.fit_transform(x_train.reshape(-1, 1))
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x_poly, y_train)
```

```python
x_poly_test = poly_reg.transform(x_test.reshape(-1,1))
```

```python
print(type(x_poly_test))
```

```python
np.array(x_poly_test)
```

```python
y_pred = lin_reg.predict(x_poly_test)
```

```python
y_pred[:5]
```

```python
from sklearn.metrics import mean_absolute_error as mae, mean_squared_error as mse, r2_score as r2s
```

```python
mae1 = mae(y_test, y_pred)
mse1 = mse(y_test, y_pred)
rscore = r2s(y_test, y_pred)
```

```python
print(f'The mae is {mae1}, the mse is {mse1}, the rscore is {rscore}')
```

```python
plt.scatter(x_test, y_test, color = 'red', s=100, alpha=0.3)
plt.scatter(x_test, y_pred, color = 'green')
plt.xlabel('Input numbers')
plt.ylabel('Squares')
plt.title('squares of inputs numbers')
plt.show()
```
