```python
col_numbers = np.arange(1, 101, 3)
col_numbers
```

Declaring the column col_numbers and creating the array of it using the numpy library as np.arrange(starts_with, end (not included), increment/decrement) 
and the printing as col_numbers

```python
col_double = col_numbers * 2
col_double
```

Multiplying each number by 2

                                                    |
                                        x_train     |  y_train
                                        ------------------------
                                                    |
                                        x_test      |  y_test 
                                                    |

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(col_numbers, col_double, test_size=0.2, random_state=42)

```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(col_numbers, col_double)
plt.xlabel('numbers')
plt.ylabel('doubles')
plt.title('numbers vs doubles')
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
lr = LinearRegression()

lr.fit(x_train.reshape(-1, 1), y_train)   
```

```python
y_pred = lr.predict(x_test.reshape(-1, 1))
```

```python
from sklearn.metrics import mean_absolute_error as mae, mean_squared_error as mse, r2_score as r2s
```

```python
mae1 = mae(y_test, y_pred)
mse1 = mse(y_test, y_pred)
r2score = r2s(y_test, y_pred)

print(f'mean_abs_error is {mae1}, mean_squared_error is {mse1}, r2score is {r2score}')
```

```python
plt.scatter(x_train, y_train, color='blue')
plt.plot(x_test, y_pred, color='red')
plt.title('train vs test')
plt.xlabel('numbers')
plt.ylabel('doubles')
```
